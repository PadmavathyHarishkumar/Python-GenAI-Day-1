```python
import pandas as pd
import numpy as np
data ={'name':['harish','padma','dev','venkhat','kumar'],'age':[29,25,20,31,68],'salary':[32000,40000,50000,60000,70000],'dept':['IT','HR','IT','Finance','HR']}
df = pd.DataFrame(data)
print(df)
```

1.Adding and removing columns

Modifying the shape of the DataFrame

```python
df_modify = df[['name','age','salary','dept']].copy()
print('Modified DataFrame:')
print(f'{df_modify}\n')
print(df_modify.columns.tolist())
```

Adding a new column

```python
df_modify['position']=['TL','TL','TL','Manager','Manager']
print('After adding position column:')
print(df_modify)
```

```python
#Removing a single column
#Method 1: Removing a single column

df_drop1 = df_modify.drop('position',axis = 1)
print('After removing position column')
print(df_drop1)
```

```python
df_modify
```

```python
#Method 2: Removing multiple columns

df_drop2 = df_modify.drop(['dept','position'],axis = 1)
print('After removing multiple columns')
print(df_drop2)
```

2.Renaming columns

Changing column names so it would be more meaningful

```python
df_rename = df[['name','age','salary','dept']].copy()
print('Original columns:')
print(f'{df_rename.columns.tolist()}\n')
print('Original DataFrame:')
print(df_rename)
print(f'{'='*40}\n')

#Changing column names
df_rename.rename(columns = { 'name' : 'emp name' , 'age' : 'emp age' , 'salary' : 'income' , 'dept' : 'area' },inplace = True)
print('After renaming columns:')
print(df_rename)
print(f'{'='*40}\n')

#After renaming
print(df_rename.columns.tolist())
print('Renamed columns')
print(df_rename)
```

3.Reading and writing data

Saving data in various formats and reading them

```python
#Saving in csv format

df[['name','age','salary','dept']].to_csv('employees.csv' , index = False)
print('Data stored to employees.csv')
print('employees.csv')
with open('employees.csv','r') as file:
    print(file.read())
```

```python
#Reading the file back in dataframe

df_read_csv = pd.read_csv(r'employees.csv')
print('Reading data from csv')
print(df_read_csv)
```

4.Exploring charts

Loading chart diagrams

```python
#Chart 1: Bar chart

import pandas as pd
import numpy as np
data ={'name':['harish','padma','dev','venkhat','kumar'],'age':[29,25,20,31,68],'salary':[32000,40000,50000,60000,70000],'dept':['IT','HR','IT','Finance','HR']}
df = pd.DataFrame(data)
print('Original DataFrame')
print(df)
print(f'{'='*40}\n')

print('Bar chart using original DataFrame')
print(df.plot(kind = 'bar'))
```

```python
#Chart 2: Pie chart

df_modify1 = df[['age','salary']].copy()
print(df_modify1)

print('Pie chart using original DataFrame')
print(df_modify1.plot(kind = 'pie',subplots = True))
```

```python
#Chart 3: Histogram

print('Original DataFrame')
print(df)
print(f'{'='*40}\n')

print('Histo chart using original DataFrame')
print(df.plot(kind = 'hist'))
```
