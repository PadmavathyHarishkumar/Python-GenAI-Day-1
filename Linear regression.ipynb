1.Single Linear Regression using manual method

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


Data = {'Tomato(kg)':[0.5,1.0,1.5,2.0,2.5,3.0,3.5,4.0,4.5,5.0,10.0],
        'Price':[20,40,60,80,100,120,140,160,180,200,400]}
df = pd.DataFrame(Data)
print('Price of tomatoes')
print(df,'\n')

#Finding linear regression
#Formula: y = mx+b
#x is the input variable
#y is the predicted variable
#b is the base value
#m is the slope

#Calculate slope value(m)
X = df['Tomato(kg)']
Y = df['Price']

#Calculating the mean values of x and y
x_mean = np.mean(X)
y_mean = np.mean(Y)

#Formula to calculate slope:
# m = np.sum(X - x_mean)(Y - y_mean)/np.sum(X - x-mean)^2
#Applying the formula

nominator = np.sum((X - x_mean)*(Y - y_mean))
denominator = np.sum((X - x_mean)**2)

m = nominator / denominator
print('Calculated slope value is :',float(round(m)))

#Calculate intercept value(b)
#Formula
#b = mx - y
#Applying the formula

b = y_mean - m*x_mean
print('\nCalculated intercept value is :',float(round(b)))

predict = 40.0*7.0 + 0.0
print('\nPredicted value of tomato for 7kg :',predict)

newx_pred = np.array([2.3,3.2,4.8,6.5,2.25,9.0])
newy_pred = m*newx_pred+b
print('\nNew predicted values are :', np.round(newy_pred).astype(int))


plt.scatter(X,Y,color = 'blue' , label = 'Data points')
plt.plot(X , m*X+b , color = 'Red' , label = 'Regression line')
plt.scatter(newx_pred, newy_pred, color='green', marker='o', s=100, label='Prediction Point')
plt.xlabel('Tomato(kg)')
plt.ylabel('Price')
plt.title('\nLinear Regression: Tomato Weight vs Price')
plt.legend()
print(plt.show())
```

2.Single Linear Regression using scikit

```python
from sklearn.linear_model import LinearRegression
df = pd.DataFrame(Data)
print('Price of tomatoes')
print(df,'\n')

X = df[['Tomato(kg)']]
Y = df['Price']

#Training the regression model
model = LinearRegression()
model.fit(X,Y)

print('Slope value is :',float(round(model.coef_[0])))
print('\nIntercept value is :',float(round(model.intercept_)))

#Predictions
y_pred = model.predict(X)

#Predictions for new values
new_x = np.array([[8.0],[8.3],[7.2],[7.8]])
sci_pred = model.predict(new_x)
print('\n',np.round(sci_pred).astype(int))

plt.scatter(X,Y,color = 'Blue' , label = 'Original Data')
plt.plot(X,y_pred,color='Red',label = 'Regression Line')
plt.scatter(new_x,sci_pred,color = 'Green',s=100,label = 'Prediction points')
plt.xlabel('Tomato(kg)')
plt.ylabel('Price')
plt.title('\nLinear Regression: Tomato Weight vs Price')
plt.legend()
print(plt.show)
```
