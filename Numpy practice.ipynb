1. Creating Arrays using numpy and accessing their elements and finding their attributes

```python
import numpy as np
#Creating a 1D Array
arr1D = np.array([10,20,30,40,50], dtype = 'float32')
print('1D Array created using numpy')
print('='*40)
print(f'{arr1D}\n')
print('Third element of the array is :',arr1D[2])
print('Fifth element of the array is :',arr1D[4])
print('First element of the array is :',f'{arr1D[0]}\n')
print('Shape of the created 1D array is : ',f'{arr1D.shape}\n')
print('Size of the created 1D array is : ',f'{arr1D.size}\n')
print('Number of dimension in the created 1D array is : ',f'{arr1D.ndim}\n')
print('Type of the created 1D array is : ',f'{type(arr1D)}\n')
print('Datatype of the created 1D array is :',f'{arr1D.dtype}\n')
```

```python
#Creating a 2D Array
arr2D = np.array([[1,'3',5],[2,4,9]])
print('2D Array created using numpy')
print('='*40)
print(f'{arr2D}\n')
print('Element in first row and first column is :',arr2D[1,1])
print('Element in Zero row and Second column is :',arr2D[0,2])
print('Element in first row and Second column is :',f'{arr2D[1,2]}\n')
print('Shape of the created 2D array is : ',f'{arr2D.shape}\n')
print('Size of the created 2D array is : ',f'{arr2D.size}\n')
print('Number of dimension in the created 2D array is : ',f'{arr2D.ndim}\n')
print('Type of the created 2D array is : ',f'{type(arr2D)}\n')
print('Datatype of the created 2D array is :',f'{arr2D.dtype}\n')
```

```python
#Creating a 3D Array
arr3D = np.array([[[1,3,5],[2,4,6]] , [[3,6,9],[2,6,8]] , [[9,12,15],[12,14,16]]])
print('3D Array created using numpy')
print('='*40)
print(f'{arr3D}\n')
print('Element in first row and Second column in matrix zero is :',arr3D[0,1,2])
print('Element in first row and Second column in matrix one is :',arr3D[1,1,2])
print('Element in first row and Second column in matrix two is :',f'{arr3D[2,1,2]}\n')
print('Shape of the created 3D array is : ',f'{arr3D.shape}\n')
print('Size of the created 3D array is : ',f'{arr3D.size}\n')
print('Number of dimension in the created 3D array is : ',f'{arr3D.ndim}\n')
print('Type of the created 3D array is : ',f'{type(arr3D)}\n')
print('Datatype of the given 3D array is : 'f'{arr3D.dtype}\n')
```

2.Creating a new 2D array and slicing their elements row wise and column wise

```python
import numpy as np
years = np.array([[2000,2001,2002,2003],[2004,2005,2006,2007],[2008,2009,2010,2011],[2012,2013,2014,2015]])
print('Years in a 4x4 numpy array is')
print('='*40)
print(f'{years}\n')
```

```python
print('Original year array')
print('='*40)
print(f'{years}\n')
print('==> Years in third row index is : 'f'{years[3:]}\n')
print('==> All years till second row index is :\n',f'{years[:3]}\n')
print('==> Years in third column index is : 'f'{years[:,3]}\n')
print('==> All years till second column index is :\n',f'{years[:,0:3]}\n')
print('==> Years in the middle of the second row is : 'f'{years[2,1:3]}\n')
print('==> Years in the middle of the first and second row is :\n 'f'{years[1:3,1:3]}\n')
```

3.Converting arrays 2D to 1D and 1D to 2D

```python
#Method 1:-Reshape
arr = np.array([2000,2004,2008,2012,2016,2020])
print('Series of years : 'f'{arr}\n')
leap_years = arr.reshape(2,3)
print('After Reshaping the series will look like')
print('='*40)
print(f'{leap_years}\n')
```

```python
#Method 2:-Flatten
arr1 = np.array([[2000,2001,2002,2003],[2004,2005,2006,2007],[2008,2009,2010,2011],[2012,2013,2014,2015]])
print('Table will look like :\n',f'{arr1}\n')
year_series = arr1.flatten()
year_series[0]=1990
print('After flattening table the result will be')
print('='*40)
print(f'{year_series}\n')
print(f'{arr1}\n')

#Note:- flatten will not affect the original array it creates a copy of an array and returns it so original will not be affected
```

```python
#Method 3:-Ravel
arr2 = np.array([[2000,2001,2002,2003],[2004,2005,2006,2007],[2008,2009,2010,2011],[2012,2013,2014,2015]])
print('Table will look like :\n',f'{arr2}\n')
years_series = arr2.ravel()
years_series[0] = 1990
print('After ravel the result will be')
print('='*40)
print(f'{years_series}\n')
print(f'{arr2}\n')

#Note:- Ravel will not create a copy and return, it directly make changes in original that may affect the original
```

4.Performing various operations in numpy array using Arithmetic operators

```python
a = np.array([2,4,6])
b = np.array([1,4,3])

print('Result after adding two arrays : ',(a+b))
print('Result after subtracting two arrays : ',(a-b))
print('Result after multiplying two arrays : ',(a*b))
print('Result after dividing two arrays : ',(a/b))
print('Result after Integer division of two arrays : ',(a//b))
print('Modulo after dividing two arrays : ',(a%b))
print('Value of a with power of b is : ',(a**b))
print('Finding a is equal to b : ',(a==b))
```

5.Broadcasting

```python
car_price = np.array([[1700000,2000000,2400000],[1800000,2300000,2500000]])
offer_price = np.array([50000,100000,70000])
print('Original price of car is')
print('='*40)
print(f'{car_price}\n')
print('Offer price on Diwali is')
print('='*40)
print(f'{offer_price}\n')
print('Final onroad price of car is')
print('='*40)
print(f'{car_price - offer_price}\n')
```

6.Aggregate functions on numpy array

```python
car_price = np.array([[1700000,2000000,2400000],[1800000,2300000,2500000]])
print('Original price of car is')
print('='*40)
print(f'{car_price}\n')
print('Result after using aggregate functions')
print('='*40)
print('Total value of all cars : ',car_price.sum())
print('Mean value of cars : ',car_price.mean())
print('Median value of cars : ',np.median(car_price))
print('Max value of cars : ',car_price.max())
print('Min value of cars : ',car_price.min())
print('Standard variation is : ',car_price.std())
print('Price variance is : ',car_price.var())
print('Sum of prices in rows : ' ,car_price.sum(axis=1))
print('Sum of prices in columns : ' ,car_price.sum(axis=0))
```

7.Matrix operations on numpy array

```python
#Transpose of matrix
mat = np.array([[1,3,5],[2,4,1]])
print('Original matrix')
print('='*40)
print(f'{mat}\n')
mat_trans = mat.transpose()
print('Transpose matrix')
print('='*40)
print(f'{mat_trans}\n')
```

```python
#Dot matrix
#Method 1:-Row and column of same size
mat1 = np.array([[1,3,5],[2,4,1],[1,3,5]])
mat2 = np.array([[1,2,3],[3,2,1],[3,2,1]])
print('Original matrix is')
print('='*40)
print(f'{mat1}\n')
print(f'{mat2}\n')
print('Final value after calculating using Dot matrix is')
print('='*50)
print(np.dot(mat1,mat2))
```

```python
#Method 2:-Row and column of different size
mat1 = np.array([[1,3,5],[2,4,1]])
mat2 = np.array([[1,2,3],[3,2,1]])
print('Original matrix is')
print('='*40)
print(f'{mat1}\n')
print(f'{mat2}\n')
print('Final value after calculating using Dot matrix is')
print('='*50)
print(np.dot(mat1,mat2.T))
```

8.Sorting operations on numpy array

```python
#Method 1:-Sorting
marks = np.array([90,60,75,88,91])
print('Original list of marks : ' ,f'{marks}\n')
print('Sorted list of marks : ', f'{np.sort(marks)}\n')
```

```python
#Method 2:-Argsort
marks = np.array([90,60,75,88,91])
print('Original list of marks : ' ,f'{marks}\n')
print('ArgSorted list : ', f'{np.argsort(marks)}\n')
names = np.array(['Harish','Padma','Dev','Shakthi','Dharan'])
print('list of names : ' ,f'{names}\n')
marks_in_order = np.argsort(marks)
print('Sorted list of marks : ', f'{np.sort(marks)}\n')
print('Names sorted using index value of marks : ',names[marks_in_order])
```

```python
#Method 3:-Conditional sort
scores = np.array([20,55,30,70,90])
print('Original values : ' f'{scores}\n')
print('Scores above 50 : ' f'{np.where(scores > 50)}\n')
print('Result of exams based on marks')
print('='*40)
print(f'{scores}\n')
print(f'{np.where(scores > 50 ,'PASS','FAIL')}\n')
```

9.Concatenating numpy arrays

```python
#Method 1:-Concatenate on rows
mat1 = np.array([[1,3,5],[2,4,1]])
mat2 = np.array([[1,2,3],[3,2,1]])
print('Concatenating on rows')
print('='*40)
print(f'{np.concatenate([mat1,mat2])}\n')

#Method 2:-Concatenate on columns
mat1 = np.array([[1,3,5],[2,4,1]])
mat2 = np.array([[1,2,3],[3,2,1]])
print('Concatenating on columns')
print('='*40)
print(f'{np.concatenate([mat1,mat2],axis=1)}\n')
```
